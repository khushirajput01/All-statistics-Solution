#VARIANCE

1.Sample Variance

2.Pop Variance
using Numpy

```python
import numpy as np
data = [2,4,4,5,6,7,8,9]
Sample_Variance = np.var(data, ddof = 1)
Pop_Variance = np.var(data, ddof = 0)

print("Sample varience is", Sample_Variance)
print("Population varince is", Pop_Variance)
```

2. Variance using Pandas

```python
import pandas as pd
data = [2,5,6,67,2,5,4,9,14,2,10]
Series = pd.Series(data)

sample_Var = Series.var()
pop_var = Series.var(ddof = 0)

print("sample var is", sample_Var)
print("pop var is", pop_var)
```

Varience using SCIPY

```python
from scipy  import stats
data = [7,8,9,1,8,5,2,6,4,7]
sample_variance = stats.tvar(data)
pop_variance = stats.tvar(data, ddof=0)
print("sample variance is", sample_variance)
print("Pop variance is", pop_variance)
```

#STANDARD DEVIATION

1. using NUMPY

```python
import numpy as np
data = [60,65,70,75,80]
sample_std_dev = np.std(data, ddof=1)
print("the sample Standard deviantion", sample_std_dev)

pop_std_dev = np.std(data, ddof=0)
print("the pop Standard deviantion is", pop_std_dev )
```

2. using PANDAS

```python
import pandas as pd
data = [2,5,6,67,2,5,4,9,14,2,10]
Series = pd.Series(data)

sample_std_deviation = Series.std()
print("The sample std dev is", sample_std_deviation)

pop_std_deviation = Series.std(ddof=0)
print("The pop std dev is", pop_std_deviation)
```

3. using SCIPY

```python
from scipy import stats
data = [2,34,7,5,8,2,1]
sample_std = stats.tstd(data)
print("sample std is", sample_std)

pop_std = stats.tstd(data, ddof=0)
print("the pop std is", pop_std)
```

#Data Representation

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#create a figure of axes
fig, axs = plt.subplots(3, 2, figsize=(14, 12))
fig.suptitle("histograms of KDE of various Distribution", fontsize=16)

#1. normal distribution (Gaussian distribution)
data_normal = np.random.normal(loc=0, scale=1, size=1000,) #mean, std dev=1
sns.histplot(data_normal, kde=True, ax=axs[0, 0], color='blue', stat='density', bins=30)
axs[0, 0].set_title('normal distrubuter')
axs[0, 0].grid(True)

#2. Left Skewed
data_left_skewed=-np.random.exponential(scale=1, size=1000)

#Left Skewed
data_right_skewed=np.random.exponential(scale=1, size=1000)

#3. Right Skewed Histogram
sns.histplot(data_right_skewed, kde=True, ax=axs[0,1], color='Red', stat='density', bins=30, label='Right skewed')

#Left skewed Histogram
sns.histplot(data_left_skewed, kde=True, ax=axs[0,1], color='green', stat='density', bins=30, label='Left skewed')

axs[0, 1].set_title('skewed Distributin')
axs[0, 1].legend()
axs[0, 1].grid(True)

#4. Uniform Distribution
data_uniform = np.random.uniform(low=0, high=1, size=1000)

#uniform distribution b/w 0 and 1
sns.histplot(data_uniform, kde=True, ax=axs[1, 0], color ='purple', stat='density', bins=30)

axs[1,0].set_title('uniform distribution')
axs=[1,0].grid(True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a figure of axes (3 rows, 2 columns)
fig, axs = plt.subplots(3, 2, figsize=(14, 12))
fig.suptitle("Histograms and KDE of Various Distributions", fontsize=16)

# 1. Normal Distribution (Top-Left)
data_normal = np.random.normal(loc=0, scale=1, size=1000)
sns.histplot(data_normal, kde=True, ax=axs[0, 0], color='blue', stat='density', bins=30)
axs[0, 0].set_title('Normal Distribution')
axs[0, 0].grid(True)

# Data Generation for Skewed Distributions
data_right_skewed = np.random.exponential(scale=1, size=1000)
# Corrected Left Skew: Max value mein se subtract karke data ko flip kiya
data_left_skewed = max(data_right_skewed) - data_right_skewed

# 2. Right Skewed Histogram (Top-Right)
sns.histplot(data_right_skewed, kde=True, ax=axs[0, 1], color='red', stat='density', bins=30)
axs[0, 1].set_title('Right Skewed (Positive Skew)')
axs[0, 1].grid(True)

# 3. Left Skewed Histogram (Middle-Left -> axs[1, 0])
sns.histplot(data_left_skewed, kde=True, ax=axs[1, 0], color='green', stat='density', bins=30)
axs[1, 0].set_title('Left Skewed (Negative Skew)')
axs[1, 0].grid(True)

# Baki bache hue khali 3 subplots ko hide karne ke liye (Sunder dikhne ke liye)
axs[1, 1].axis('off')
axs[2, 0].axis('off')
axs[2, 1].axis('off')

plt.tight_layout()
plt.show()
```
