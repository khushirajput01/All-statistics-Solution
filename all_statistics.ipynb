#CALCULATING MEAN

1. using mannual Calculation'


```python
population = [150,160,170,165,155]
sample = [150,170,165]

pop_mean = sum(population)/len(population)
sample_mean = sum(sample)/len(sample)
print("sum of population mean is", pop_mean)
print("sum of sample mean is", sample_mean)
```

2. Using NUMPY

```python
import numpy as np
data = [85,90,78,92,88]
airthemetic_mean = np.mean(data)
print("the mean is", airthemetic_mean)
```

```python
import numpy as np
data = [40,50,70,20]
airthemetic_mean = np.mean(data)
print("mean is", airthemetic_mean)
```

3. Using stastics

```python
import statistics as stats
data = [40,50,70,20]
mean_stats = stats.mean(data)
print("mean is", mean_stats)
```

4. Mean using dataset (SEABORN)


```python
import seaborn as sns
import numpy as np

iris = sns.load_dataset('iris')
iris.head()

mean_sepal_length = np.mean(iris['sepal_length'])
print("The mean of sepal length is", mean_sepal_length)
```

#Median


1. calculate Median using Numpy


```python

import numpy as np
data  = [20,22,24,26,28,100]
median = np.median(data)
print("Then Median of data is", median)
```

2. Median using Statistics


```python
import statistics as stats
data = [1,41,54,5,51,56]
median = stats.median(data)
print("median", median)
```
